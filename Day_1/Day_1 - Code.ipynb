# **Printing**

#### **Printing the classsic 'Hello World!' to the console!**

The word **'print'** is the **function** telling the computer to simply **output** the text inside the brackets. There are also double quotes surrounding the text in order to indicate to the computer that the text is a **string**, meaning it's just a sequence of characters, not a **command** for the computer to run. The quotes mark exactly where that string starts and ends.

```python
print("Hello world!")
```

The **output** for this line of code would be:

```python
Hello world!
```

#### **Printing Practice - Udemy Exercise**

In this **exercise**, we have to turn a set of instructions to bake white bread into a **program** for the computer **output** the instructions into the **console**.

**BEFORE:**

```python
1. Mix 500g of Flour, 10g Yeast and 300ml Water in a bowl.
2. Knead the dough for 10 minutes.
3. Add 3g of Salt.
4. Leave to rise for 2 hours.
5. Bake at 200 degrees C for 30 minutes.
```

**AFTER:**

```python
print("1. Mix 500g of Flour, 10g Yeast and 300ml Water in a bowl.")
print("2. Knead the dough for 10 minutes.")
print("3. Add 3g of Salt.")
print("4. Leave to rise for 2 hours.")
print("5. Bake at 200 degrees C for 30 minutes.")
```

**OUTPUT:**

```python
1. Mix 500g of Flour, 10g Yeast and 300ml Water in a bowl.
2. Knead the dough for 10 minutes.
3. Add 3g of Salt.
4. Leave to rise for 2 hours.
5. Bake at 200 degrees C for 30 minutes.
```
